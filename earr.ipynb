```python
import cv2
import numpy as np

min_YCrCb = np.array([0,133,77], np.uint8)
max_YCrCb = np.array([235,173,127], np.uint8)

# Use existing image
image = cv2.imread("000_back_ear.jpg")

if image is None:
    print("Image not found")
else:

    imageYCrCb = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2YCrCb
    )

    skinRegionYCrCb = cv2.inRange(
        imageYCrCb,
        min_YCrCb,
        max_YCrCb
    )

    skinYCrCb = cv2.bitwise_and(
        image,
        image,
        mask=skinRegionYCrCb
    )

    cv2.imwrite(
        "indexycrcb.png",
        np.hstack([image, skinYCrCb])
    )

    print("Skin detection complete")
```
